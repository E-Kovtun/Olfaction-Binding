# Full_full LoRaX site-level MIL / attention screen

Plot-only notebook for mixed-granularity molecule-site GIN x mean protein ESM-1b experiments.
Six methods are compared: `boost_gin_mean` (XGBoost control on mean-pooled embeddings), three
permutation-invariant MIL pooling heads (`mil_max`, `mil_noisy_or`, `mil_lse`), and two learned
attention heads over molecule sites (`attention_mil`, `attention_excess_max`).

Data protocol:

- transductive: genuine LoRaX `rand_split_1..5`;
- inductive_molecule: cold-molecule split seeds `42..46` from the fold-1 full pair pool;
- molecule keys: LoRaX SMILES -> InChIKey via `data/processed/molecules/lorax_smiles_to_inchikey.csv`;
- molecule embeddings: `gin_supervised_contextpred_all_m2or[_per_atom].npz`;
- protein embeddings: LoRaX `esm1b_650m_mean.npz`.

Run script:

```bash
bash legacy/scripts/queues/run_full_full_site_mil_attention.sh
```

````python
from pathlib import Path
import math

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t as student_t

plt.rcParams.update({
    "figure.figsize": (11, 4),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
COLORS = {
    "transductive": "#4C78A8",
    "inductive_molecule": "#F58518",
}
MODEL_ORDER = ["boost_gin_mean", "mil_max", "mil_noisy_or", "mil_lse", "attention_mil", "attention_excess_max"]

ROOT = Path("..").resolve().parent if Path.cwd().name == "baselines_research" else Path.cwd()
RESULT_DIR = ROOT / "results" / "attention" / "full_full" / "site_mil"
METRICS = ["AUROC", "AUPRC", "MCC", "F1"]
metrics_path = RESULT_DIR / "metrics.csv"
print("results:", RESULT_DIR)
print("metrics exists:", metrics_path.exists())
````

````python
if metrics_path.exists():
    runs = pd.read_csv(metrics_path)
else:
    runs = pd.DataFrame(columns=["regime", "repeat", "model", *METRICS])

display_cols = ["regime", "repeat", "model", "threshold", "n_train", "n_val", "n_test", "test_pos", *METRICS]
available_cols = [c for c in display_cols if c in runs.columns]
print(f"loaded {len(runs)} runs")
if len(runs):
    display(runs[available_cols].sort_values(["regime", "model", "repeat"]).head(20))
else:
    print("No runs yet. Launch legacy/scripts/queues/run_full_full_site_mil_attention.sh on the server.")
````

````python
def ci95(x):
    x = np.asarray(pd.Series(x).dropna(), dtype=float)
    n = len(x)
    if n == 0:
        return pd.Series({"mean": np.nan, "sd": np.nan, "ci_low": np.nan, "ci_high": np.nan, "n": 0})
    mean = float(x.mean())
    sd = float(x.std(ddof=1)) if n > 1 else np.nan
    half = float(student_t.ppf(0.975, n - 1) * sd / math.sqrt(n)) if n > 1 else np.nan
    return pd.Series({"mean": mean, "sd": sd, "ci_low": mean - half if n > 1 else np.nan, "ci_high": mean + half if n > 1 else np.nan, "n": n})

if len(runs):
    summary = (
        runs.groupby(["regime", "model"])[METRICS]
        .agg(lambda s: ci95(s)["mean"])
        .reset_index()
        .sort_values(["regime", "AUPRC"], ascending=[True, False])
    )
    ci_rows = []
    for (regime, model), g in runs.groupby(["regime", "model"]):
        for metric in METRICS:
            row = ci95(g[metric]).to_dict()
            row.update({"regime": regime, "model": model, "metric": metric})
            ci_rows.append(row)
    ci = pd.DataFrame(ci_rows)
    display(summary)
else:
    summary = pd.DataFrame()
    ci = pd.DataFrame()
````

````python
# Grouped bar chart, one panel per metric: model on the x-axis, regime as bar color,
# 95% CI (Student-t over repeats) as error bars.
if len(ci):
    models_present = [m for m in MODEL_ORDER if m in ci["model"].unique()] or sorted(ci["model"].unique())
    regimes_present = [r for r in COLORS if r in ci["regime"].unique()] or sorted(ci["regime"].unique())

    fig, axes = plt.subplots(2, 2, figsize=(14, 8))
    axes = axes.ravel()
    x = np.arange(len(models_present))
    width = 0.8 / max(len(regimes_present), 1)

    for ax, metric in zip(axes, METRICS):
        d = ci[ci.metric == metric]
        for i, regime in enumerate(regimes_present):
            g = d[d.regime == regime].set_index("model").reindex(models_present)
            offset = (i - (len(regimes_present) - 1) / 2) * width
            err_low = (g["mean"] - g["ci_low"]).to_numpy()
            err_high = (g["ci_high"] - g["mean"]).to_numpy()
            err = np.vstack([np.nan_to_num(err_low), np.nan_to_num(err_high)])
            bars = ax.bar(x + offset, g["mean"], width=width, yerr=err, capsize=3,
                           color=COLORS.get(regime, None), alpha=0.85, label=regime)
            ax.bar_label(bars, fmt="%.3f", fontsize=7, padding=2, rotation=90)
        ax.set_title(metric)
        ax.set_xticks(x)
        ax.set_xticklabels(models_present, rotation=30, ha="right")
        ax.set_ylim(0, 1)
        ax.set_ylabel(metric)

    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="upper center", ncol=len(regimes_present), frameon=False, bbox_to_anchor=(0.5, 1.04))
    fig.suptitle("Full_full LoRaX site-MIL/attention screen: mean +/- 95% CI over repeats", y=1.08)
    fig.tight_layout()
    plt.show()
else:
    print("No CI plot yet.")
````

````python
if len(runs):
    coverage_cols = [c for c in runs.columns if c.endswith("_rows_after") or c in ["regime", "repeat", "model", "n_train", "n_val", "n_test"]]
    coverage = runs[coverage_cols].drop_duplicates(["regime", "repeat"]).sort_values(["regime", "repeat"])
    display(coverage)
else:
    print("No coverage table yet.")
````

````python
# Test score distributions, one row per model, one column per regime.
# For each (regime, model) we take the repeat with the highest AUPRC (not the single
# best model overall -- otherwise the boosting control alone would dominate the grid).
pred_dir = RESULT_DIR / "predictions"
if len(runs) and pred_dir.exists():
    best = runs.sort_values("AUPRC", ascending=False).groupby(["regime", "model"]).head(1)
    pred_frames = []
    for row in best.itertuples():
        run_id = row.run_id if hasattr(row, "run_id") else f"{row.regime}_rep{row.repeat}_{row.model}_seed{row.seed}_boost{row.boost_seed}"
        path = pred_dir / f"test_{run_id}.csv"
        if path.exists():
            df = pd.read_csv(path)
            df["regime"] = row.regime
            df["model"] = row.model
            df["run_id"] = run_id
            pred_frames.append(df)

    if pred_frames:
        pred = pd.concat(pred_frames, ignore_index=True)
        row_models = [m for m in MODEL_ORDER if m in pred["model"].unique()] or sorted(pred["model"].unique())
        col_regimes = [r for r in COLORS if r in pred["regime"].unique()] or sorted(pred["regime"].unique())

        fig, axes = plt.subplots(len(row_models), len(col_regimes),
                                  figsize=(5.2 * len(col_regimes), 3.0 * len(row_models)),
                                  squeeze=False)
        for i, model in enumerate(row_models):
            for j, regime in enumerate(col_regimes):
                ax = axes[i][j]
                g = pred[(pred.model == model) & (pred.regime == regime)]
                if g.empty:
                    ax.axis("off")
                    continue
                ax.hist(g.loc[g.y == 0, "p"], bins=40, density=True, alpha=0.55,
                        color="#4C78A8", label="negative")
                ax.hist(g.loc[g.y == 1, "p"], bins=40, density=True, alpha=0.55,
                        color="#D62728", label="positive")
                if i == 0:
                    ax.set_title(regime)
                if j == 0:
                    ax.set_ylabel(model, fontsize=9)
                if i == len(row_models) - 1:
                    ax.set_xlabel("p")
                if i == 0 and j == 0:
                    ax.legend(frameon=False, fontsize=8)
        fig.suptitle("Best-by-AUPRC test score distributions (all methods)", y=1.02)
        fig.tight_layout()
        plt.show()
    else:
        print("Prediction files for best runs were not found yet.")
else:
    print("No predictions yet.")
````

## Attention-weight diagnostics

`attention_mil` and `attention_excess_max` are the only two methods that compute a genuine
normalized attention distribution over molecule sites (a softmax); their max weight is a
meaningful "how concentrated is the model's focus" statistic. `mil_max` / `mil_noisy_or` /
`mil_lse` score sites individually and pool them (max, noisy-OR, log-sum-exp) but never
produce a normalized weight to plot here, so they are intentionally left out of this panel
(they already appear in the prediction-distribution grid above).

For `attention_excess_max` the final prediction *is* the excess max-attention score (see the
`ExcessAttentionMax.score` definition), so its row here should closely mirror its row in the
prediction-distribution grid above -- that overlap is expected and is itself a useful sanity
check. `attention_mil` uses a learned head on top of the attention-weighted bag, so its
attention weights and its final prediction can genuinely diverge.

Weights are recomputed from the saved checkpoints (`checkpoints/{run_id}.pt`) applied to the
saved test pairs, one molecule/protein pair at a time -- nothing is retrained.

````python
# Recompute per-pair max attention weight from the saved checkpoints for the two
# genuine attention models, for the same (regime, model) runs selected above.
import argparse
import json
import sys

import torch

TRAIN_DIR = ROOT / "scripts" / "modeling" / "train"
if str(TRAIN_DIR) not in sys.path:
    sys.path.insert(0, str(TRAIN_DIR))
import train_full_full_site_mil_attention as ffs

ATTENTION_MODELS = ["attention_mil", "attention_excess_max"]

def attention_mil_weights(model, protein, sites, mask):
    p = model.protein(protein)
    s = model.site(sites)
    p_rep = p.unsqueeze(1).expand_as(s)
    a_in = torch.cat([s, p_rep], dim=-1)
    a = model.attn_w(model.attn_v(a_in) * model.attn_u(a_in)).squeeze(-1).masked_fill(mask, -torch.inf)
    return torch.softmax(a, dim=1)

def load_max_attention(run_id, model_name):
    run_json = RESULT_DIR / "runs" / run_id / "metrics.json"
    ckpt_path = RESULT_DIR / "checkpoints" / f"{run_id}.pt"
    pred_path = RESULT_DIR / "predictions" / f"test_{run_id}.csv"
    if not (run_json.exists() and ckpt_path.exists() and pred_path.exists()):
        return None

    with open(run_json, encoding="utf-8") as f:
        cfg = json.load(f)["config"]
    args = argparse.Namespace(**cfg)

    proteins = ffs.load_npz_dict(ROOT / cfg["protein"])
    sites = ffs.load_npz_dict(ROOT / cfg["molecule_sites"])

    model = ffs.build_model(model_name, args)
    ckpt = torch.load(ckpt_path, map_location="cpu")
    model.load_state_dict(ckpt["state_dict"])
    model.eval()

    df = pd.read_csv(pred_path)
    max_w = np.full(len(df), np.nan, dtype=np.float32)
    with torch.no_grad():
        for i, row in enumerate(df.itertuples()):
            if row.protein not in proteins or row.inchikey not in sites:
                continue
            prot = torch.as_tensor(proteins[row.protein], dtype=torch.float32).unsqueeze(0)
            site = torch.as_tensor(sites[row.inchikey], dtype=torch.float32).unsqueeze(0)
            mask = torch.zeros(1, site.shape[1], dtype=torch.bool)
            if model_name == "attention_excess_max":
                w = model.weights(prot, site, mask)
            else:
                w = attention_mil_weights(model, prot, site, mask)
            max_w[i] = float(w.max().item())
    out = df[["y"]].copy()
    out["max_attention"] = max_w
    out["model"] = model_name
    return out

if len(runs) and (RESULT_DIR / "checkpoints").exists():
    best_attn = runs[runs["model"].isin(ATTENTION_MODELS)].sort_values("AUPRC", ascending=False).groupby(["regime", "model"]).head(1)
    attn_frames = []
    for row in best_attn.itertuples():
        run_id = row.run_id if hasattr(row, "run_id") else f"{row.regime}_rep{row.repeat}_{row.model}_seed{row.seed}_boost{row.boost_seed}"
        out = load_max_attention(run_id, row.model)
        if out is not None:
            out["regime"] = row.regime
            attn_frames.append(out)

    if attn_frames:
        attn = pd.concat(attn_frames, ignore_index=True)
        row_models = [m for m in ATTENTION_MODELS if m in attn["model"].unique()]
        col_regimes = [r for r in COLORS if r in attn["regime"].unique()] or sorted(attn["regime"].unique())

        fig, axes = plt.subplots(len(row_models), len(col_regimes),
                                  figsize=(5.2 * len(col_regimes), 3.0 * len(row_models)),
                                  squeeze=False)
        for i, model in enumerate(row_models):
            for j, regime in enumerate(col_regimes):
                ax = axes[i][j]
                g = attn[(attn.model == model) & (attn.regime == regime)].dropna(subset=["max_attention"])
                if g.empty:
                    ax.axis("off")
                    continue
                ax.hist(g.loc[g.y == 0, "max_attention"], bins=40, density=True, alpha=0.55,
                        color="#4C78A8", label="negative")
                ax.hist(g.loc[g.y == 1, "max_attention"], bins=40, density=True, alpha=0.55,
                        color="#D62728", label="positive")
                if i == 0:
                    ax.set_title(regime)
                if j == 0:
                    ax.set_ylabel(model, fontsize=9)
                if i == len(row_models) - 1:
                    ax.set_xlabel("max attention weight")
                if i == 0 and j == 0:
                    ax.legend(frameon=False, fontsize=8)
        fig.suptitle("Best-by-AUPRC max attention weight (attention_mil, attention_excess_max)", y=1.02)
        fig.tight_layout()
        plt.show()
    else:
        print("No attention checkpoints/predictions found yet.")
else:
    print("No checkpoints yet.")
````
